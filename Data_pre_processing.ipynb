* NOW THERE IS NULL VALUES IN SOLAR.R -HERE WE USE MEAN TI FILL NULL VALUES BECAUSE IT ONLY HAS 7 NULL VALUES SO IT DOES NOT IMPACT THE ORIGINAL DATASET HAS THAT MUCH

```python
weather_data[weather_data["Solar.R"].isna()]
```

```python
weather_data.groupby("Weather")["Solar.R"].mean().round()
```

```python
weather_data["Solar.R"]= weather_data.groupby("Weather")["Solar.R"].transform(lambda x:x.fillna(x.mean().round()))
```

#### Using One Hot Encoder 

* This can be achieved using any of the following 2 Libraries

* pandas - pd.get_dummies()

* sklearn - OneHotEncoder()

* USING PANDAS

```python
weather_data_copy_2 = weather_data.copy()
```

```python
weather_data_copy_2 = pd.get_dummies(data=weather_data_copy_2,columns=["Weather"],dtype="int") #using this it creates a 3 different columns for each output values
weather_data_copy_2
```

* USING ONEHOTENCODER SKLEARN

```python
weather_data_copy_3 = weather_data.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
 one_encoder = OneHotEncoder()
```

```python
one_encoder.fit_transform(weather_data_copy_3[["Weather"]]) #IT returns the compressed sparse row matrix not a regular array this is a default behaviour of the onehotencoder
```

```python
one_encoder=OneHotEncoder(sparse_output=False)
```

```python
encoded = one_encoder.fit_transform(weather_data_copy_3[["Weather"]])
```

```python
weather_data_copy_3
```

```python
encoded_df = pd.DataFrame(encoded,columns=one_encoder.get_feature_names_out(["Weather"]))
```

```python
weather_data_copy_3= pd.concat([weather_data_copy_3.reset_index(drop=True),encoded_df],axis=1)
```

```python
weather_data_copy_3.head()
```

```python
weather_data_copy_3  = weather_data_copy_3.drop(columns=["Weather"])
```

# HOW TO CHOOSE BTW THE LABEL ENCODER AND ONE HOT ENCODER

* **Input Feature**
* Based on whether it is Parametric or a Non-Parametric Model we will be choosing
* For parametric model we chose - OneHotEncoder
* For Non-Parametric Model we chose - LabelEncoder

* **Output Feature**
* Always the output features must be LabelEncoder
